```python
# Cell 1 - get job parameters
catalog = dbutils.widgets.get("catalog")
schema  = dbutils.widgets.get("schema")
spark.sql(f"USE CATALOG {catalog}")
spark.sql(f"USE SCHEMA {schema}")

# Cell 2 - create and write data
rows = [
    (1, "Apples",    120),
    (2, "Bananas",    85),
    (3, "Cherries",  200),
    (4, "Dates",     310),
    (5, "Elderberry", 45),
    (6, "Figs",      175),
    (7, "Grapes",    230),
    (8, "Honeydew",   90),
]
df = spark.createDataFrame(rows, ["id", "product", "units_sold"])
df.write.mode("overwrite").saveAsTable("sales_data")
print(f"Written {df.count()} rows to sales_data table")
df.show()
```
