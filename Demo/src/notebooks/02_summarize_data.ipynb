```python
# Cell 1 - get job parameters
catalog = dbutils.widgets.get("catalog")
schema  = dbutils.widgets.get("schema")
spark.sql(f"USE CATALOG {catalog}")
spark.sql(f"USE SCHEMA {schema}")

# Cell 2 - read and summarize
from pyspark.sql import functions as F

df = spark.read.table("sales_data")
summary = df.agg(
    F.count("units_sold").alias("total_products"),
    F.sum("units_sold").alias("total_units"),
    F.avg("units_sold").alias("avg_units"),
    F.min("units_sold").alias("min_units"),
    F.max("units_sold").alias("max_units")
)
print("=== Sales Summary ===")
summary.show()
df.orderBy(F.desc("units_sold")).show()
```
